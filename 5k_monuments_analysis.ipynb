### 2. Importation du jeu de données sur les monuments et analyse

```python
path = "proj_adad.xlsx"

small_dataset = pd.read_excel(path, header=0)
```

```python
#@title Code — Sélectionner les colonnes
# df = small_dataset.loc[:, ('User_ID', 'Age', 'Gender')]

print(f'Nombre total de lignes : {len(small_dataset.index)}\n\n')
small_dataset.head()
```

### 3. Data pre-processing

```python
# pre-filter (data pre-processing)

small_dataset_pre_filter = small_dataset[
    (small_dataset["Latitude"].between(42.8, 45.2)) &
    (small_dataset["Longitude"].between(4.2, 7.8))
]
```

```python
def precise_coords(x):
    s = str(x)
    return len(s.split(".")[1]) if "." in s else 0
```

```python
mask_keep_precise = (
    small_dataset_pre_filter["Latitude"].apply(precise_coords) >= 3
) & (
    small_dataset_pre_filter["Longitude"].apply(precise_coords) >= 3
)

df = small_dataset_pre_filter[mask_keep_precise]
```

```python
# Standardize code insee (5 digits)
df["code_insee"] = df["code_insee"].astype(str).str.zfill(5)
```

Elimination de la catégorie "Site patrimonial remarquable" sur la base de sa définition (interessante pour l'état, pas pour un touriste).

```python
df = df[df["Type_equipement_ou_lieu"] != "Site patrimonial remarquable"]
```

```python
colors = [
    "red", "blue", "green", "purple", "orange",
    "darkred", "lightcoral", "beige", "darkblue",
    "darkgreen", "cadetblue", "indigo",
    "white", "pink", "lightblue", "lightgreen",
    "gray", "black", "lightgray"
]

df["color"] = (
    df["Type_equipement_ou_lieu"]
      .astype("category")
      .cat.codes
      .apply(lambda i: colors[i % len(colors)])
)
```

```python
df.describe(include='all')
```

## Partie 2 - Carte des lieus qui nous interessent (19 Décembre 2025)

### 1. Ouvrire la table des gares

```python
path_gares = "gares_PACA_horaires_geo_clean.csv"

gares_dataset = pd.read_csv(path_gares, header=0)
```

```python
print(f'Nombre total de lignes : {len(gares_dataset.index)}\n\n')
# gares_dataset.head()
```

### 2. Data pre-processing

#### 2a. "Code commune" : if 4 digits ---> '0' + Code commune

```python
gares_dataset["Code commune"] = gares_dataset["Code commune"].astype(str).str.zfill(5)
gares_dataset.head()
```

#### 2b. Filtre: mantenir seulement les codes commune de PACA : 04, 05, 06, 13, 83, 84

```python
# mantenir seulement PACA (04, 05, 06, 13, 83, 84)

codes_commune = (
    gares_dataset["Code commune"]
    .loc[lambda s: s.str.startswith(("04", "05", "06", "13", "83", "84"))]
    .unique()
)
```

```python
codes_commune = codes_commune[codes_commune != "13201"] # drop Marseille (Saint-Charles)

codes_commune
```

```python
gares_dataset_filtered = gares_dataset[gares_dataset["Code commune"].isin(codes_commune)]
```

### 3. JOIN entre les deux tables, avec "df.code_insee == gares_dataset.Code commune"

```python
df_joined = df[df["code_insee"].isin(codes_commune)]
```

```python
print(f'Nombre total de lignes : {len(df_joined.index)}\n\n')
df_joined.head()
```

### 4. Nouvelle carte

```python
m = folium.Map(
    location=[43.9, 6.0], # centre PACA
    zoom_start=8
) # create map m
```

```python
# monuments_layer = folium.FeatureGroup(name="Monuments") # create monuments_layer
```

```python
from folium.plugins import MarkerCluster

# lista dei tipi unici
types = df_joined["Type_equipement_ou_lieu"].unique()

# dizionario per layer (opzionale, se vuoi manipolarli)
layers_dict = {}

for t in types:
    # crea FeatureGroup per il tipo (overlay=True)
    layer = folium.FeatureGroup(name=t, overlay=True, show=False)
    
    # usa MarkerCluster per rendere la mappa più leggibile
    cluster = MarkerCluster().add_to(layer)
    
    # filtra il dataframe per questo tipo
    subset = df_joined[df_joined["Type_equipement_ou_lieu"] == t]
    
    for _, row in subset.iterrows():
        folium.Marker(
            location=[row["Latitude"], row["Longitude"]],
            popup=f"""
            <b>{row['Nom']}</b><br>
            <br>
            Type : {row['Type_equipement_ou_lieu']}<br>
            <br>
            <b>Commune : {row['libelle_geographique']}</b>
            """,
            icon=folium.Icon(
                color=row.get("color", "blue"),
                icon="info-sign"
            )
        ).add_to(cluster)
        
    layer.add_to(m)
    layers_dict[t] = layer
```

```python
# gares_layer = folium.FeatureGroup(name="Gares ferroviaires") # create gares_layer
```

```python
gares_distinct = (
    gares_dataset_filtered
    .dropna(subset=["Latitude", "Longitude"])
    .drop_duplicates(subset=["Nom", "Latitude", "Longitude"])
)
```

```python
# crea un FeatureGroup per le stazioni
gares_layer = folium.FeatureGroup(name="Gares", overlay=True, show=True)

# opzionale: usare MarkerCluster se ci sono molte stazioni
gares_cluster = MarkerCluster().add_to(gares_layer)

# aggiungi i marker delle stazioni
for _, row in gares_distinct.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=f"<b>Gare : {row['Nom']}</b>",
        icon=folium.Icon(
            color="purple",
            icon="train",
            prefix="fa"
        )
    ).add_to(gares_cluster)

# aggiungi il layer alla mappa
gares_layer.add_to(m)
```

# Partie 3 : graph (barplot)

```python
def plot_monuments_by_type_vertical(df, comune):
    
    def single_plot(data, comune_name):
        counts = (
            data.groupby("Type_equipement_ou_lieu")
            # data.groupby("Domaine")
                .size()
                .reset_index(name="count")
                .sort_values("count", ascending=False)
        )
        
        plt.figure(figsize=(10, 5))
        ax = sns.barplot(
            data=counts,
            x="Type_equipement_ou_lieu",
            # x="Domaine",
            y="count"
        )

        max_count = counts["count"].max()
        ax.set_ylim(0, max_count * 1.15)

        # add labels on top of bars
        for p in ax.patches:
            height = p.get_height()
            ax.annotate(
                f'{int(height)}',
                (p.get_x() + p.get_width() / 2., height),
                ha='center',
                va='bottom',
                xytext=(0, 3),
                textcoords='offset points'
            )
        
        plt.title(f"Nombre de monuments par type - {comune_name}")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # cas ALL
    if isinstance(comune, str) and comune.lower() == "all":
        for comune_name in df["libelle_geographique"].dropna().unique():
            subset = df[df["libelle_geographique"] == comune_name]
            if not subset.empty:
                single_plot(subset, comune_name)
    else:
        subset = df[df["libelle_geographique"] == comune]
        if subset.empty:
            print(f"No data found for commune : {comune}")
        else:
            single_plot(subset, comune)
```

```python
def plot_monuments_by_type(df, comune):

    def single_plot(data, comune_name):
        # conteggi
        counts = (
            data.groupby("Type_equipement_ou_lieu")
                .size()
                .reset_index(name="count")
        )
    
        # ordinamento per grafico orizzontale
        counts = counts.sort_values("count", ascending=False)
    
        plt.figure(figsize=(10, 6))
        ax = sns.barplot(
            data=counts,
            y="Type_equipement_ou_lieu",
            x="count",
            orient="h"
        )
    
        # margine asse x
        max_count = counts["count"].max()
        ax.set_xlim(0, max_count * 1.15)
    
        # etichette sui valori
        for p in ax.patches:
            width = p.get_width()
            ax.annotate(
                f'{int(width)}',
                (width, p.get_y() + p.get_height() / 2),
                ha='left',
                va='center',
                xytext=(5, 0),
                textcoords='offset points'
            )
            
        plt.title(f"Nombre de monuments par type - {comune_name}")
        plt.xlabel("Type d'équipement ou lieu")
        plt.ylabel("Nombre de monuments")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # cas ALL
    if isinstance(comune, str) and comune.lower() == "all":
        for comune_name in df["libelle_geographique"].dropna().unique():
            subset = df[df["libelle_geographique"] == comune_name]
            if not subset.empty:
                single_plot(subset, comune_name)
    else:
        subset = df[df["libelle_geographique"] == comune]
        if subset.empty:
            print(f"No data found for commune : {comune}")
        else:
            single_plot(subset, comune)
```

```python
plot_monuments_by_type(df, "Cannes")
plot_monuments_by_type(df, "Menton")
plot_monuments_by_type(df, "Arles")
plot_monuments_by_type(df, "Briançon")
# plot_monuments_by_type(df, "all")
```
